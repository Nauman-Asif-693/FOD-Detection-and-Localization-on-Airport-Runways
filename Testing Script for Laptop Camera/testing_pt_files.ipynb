```python
import cv2
import torch
from ultralytics import YOLO

# Load YOLOv8 model (replace 'path_to_yolov8_nano.pt' with your model file path)
model = YOLO('yolov8n.pt')  # Your custom-trained YOLOv8 model

# Open the laptop's camera
cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("Error: Could not open camera.")
    exit()

while True:
    # Capture frame-by-frame
    ret, frame = cap.read()
    
    if not ret:
        print("Error: Could not read frame.")
        break
    
    # Perform detection
    results = model(frame)

    # Annotate the frame with detection results
    annotated_frame = results[0].plot()  # YOLOv8 automatically returns annotated images

    # Display the resulting frame
    cv2.imshow('YOLOv8 Real-time Detection', annotated_frame)
    
    # Break the loop if 'q' is pressed
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# When everything is done, release the capture
cap.release()
cv2.destroyAllWindows()
```
